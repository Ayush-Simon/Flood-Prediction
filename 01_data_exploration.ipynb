### Archive: first exploration of Bhubaneswar rainfall
The project started with Bhubaneswar. These cells are kept for the record; the study moved to New York City because no public street flood-depth data exists for Bhubaneswar. The New York work starts at the FloodNet download below.


```python
import os
import pandas as pd

print(os.getcwd())

path = "data/raw/Odisha Hourly Rainfall 2021-2025.csv"
df = pd.read_csv(path)
print(df.shape)
print(df.columns.tolist())
df.head()
```

```python
rain = "Telemetry Hourly Rainfall (mm)"

# Turn the text dates into real dates (day-month-year)
df["time"] = pd.to_datetime(df["Data Acquisition Time"], format="%d-%m-%Y %H:%M")

print("Date range:", df["time"].min(), "to", df["time"].max())
print("Stations:", df["Station"].nunique())
print("\nRainfall summary:")
print(df[rain].describe())
print("\nZero-rain rows:", (df[rain] == 0).sum())
print("Negative rows:", (df[rain] < 0).sum())
print("Repeated station + time:", df.duplicated(["Station", "time"]).sum())
```

```python
# Per-station summary
stats = df.groupby("Station")[rain].agg(
    rows="count",
    median="median",
    max="max",
    over_100=lambda s: (s > 100).sum(),
)
print(stats.sort_values("median", ascending=False).head(15))
print("\nStations with median > 50 mm:", (stats["median"] > 50).sum(), "of", len(stats))

# Look at the most suspicious station in time order
top = stats["median"].idxmax()
print("\nStation:", top)
print(df[df["Station"] == top].sort_values("time")[["time", rain]].head(30).to_string())
```

```python
import numpy as np

# Bhubaneswar city centre (approx.)
bbsr_lat, bbsr_lon = 20.2961, 85.8245

st = df.groupby("Station").agg(
    district=("District", "first"),
    lat=("Latitude", "first"),
    lon=("Longitude", "first"),
    rows=(rain, "count"),
    first=("time", "min"),
    last=("time", "max"),
    median=(rain, "median"),
)

# Straight-line distance in km (haversine formula)
lat1, lon1 = np.radians(bbsr_lat), np.radians(bbsr_lon)
lat2, lon2 = np.radians(st["lat"]), np.radians(st["lon"])
a = np.sin((lat2 - lat1) / 2)**2 + np.cos(lat1) * np.cos(lat2) * np.sin((lon2 - lon1) / 2)**2
st["km"] = 6371 * 2 * np.arcsin(np.sqrt(a))

print(st.sort_values("km").head(10).round(2).to_string())
```

```python
b = df[df["Station"] == "Balianta_1"].sort_values("time").copy()

print("Max:", b[rain].max())
print("Negatives:", (b[rain] < 0).sum())
print("Repeated timestamps:", b.duplicated(["time"]).sum())
print("...of which exact copies (same time AND value):", b.duplicated(["time", rain]).sum())

print("\nHours above 50 mm:")
print(b[b[rain] > 50][["time", rain]].to_string())

# Totals per year and month (exact copies dropped so they aren't counted twice)
bc = b.drop_duplicates(["time", rain])
print("\nPer year:")
print(bc.groupby(bc["time"].dt.year)[rain].agg(rainy_hours="count", total_mm="sum"))
print("\nPer month (mm):")
print(bc.groupby(bc["time"].dt.to_period("M"))[rain].sum().to_string())
```

```python
import os
import numpy as np
import pandas as pd
import requests

raw_dir = os.path.abspath("data/raw")
ev = pd.read_csv(f"{raw_dir}/floodnet_events.csv")
se = pd.read_csv(f"{raw_dir}/floodnet_sensors.csv")

# 1. Pick rain-driven sensors with 10+ flood events
counts = ev["Sensor ID"].value_counts()
sensors = se[(se["Tidally Influenced"] == "No") & se["Sensor ID"].isin(counts[counts >= 10].index)]
sensors = sensors.drop_duplicates("Sensor ID")[["Sensor ID", "Latitude", "Longitude"]]
print("Sensors selected:", len(sensors))

# 2. Hourly rainfall for each sensor (UTC, to match FloodNet's GMT times)
rain_frames = []
for _, s in sensors.iterrows():
    r = requests.get("https://archive-api.open-meteo.com/v1/archive", params={
        "latitude": s["Latitude"], "longitude": s["Longitude"],
        "start_date": "2020-10-01", "end_date": "2026-10-31",
        "hourly": "precipitation", "timezone": "UTC",
    }, timeout=120)
    r.raise_for_status()
    d = pd.DataFrame(r.json()["hourly"])
    d["Sensor ID"] = s["Sensor ID"]
    rain_frames.append(d)
rain = pd.concat(rain_frames, ignore_index=True)
rain.to_csv(f"{raw_dir}/nyc_rain_hourly_openmeteo.csv", index=False)
print("Rain rows:", len(rain), "| missing:", rain["precipitation"].isna().sum())

# 3. Elevation at the sensor plus 4 points 100 m away, to work out slope
terrain = []
for _, s in sensors.iterrows():
    lat, lon = s["Latitude"], s["Longitude"]
    dlat = 100 / 111320
    dlon = 100 / (111320 * np.cos(np.radians(lat)))
    lats = [lat, lat + dlat, lat - dlat, lat, lat]
    lons = [lon, lon, lon, lon + dlon, lon - dlon]
    r = requests.get("https://api.open-meteo.com/v1/elevation", params={
        "latitude": ",".join(map(str, lats)), "longitude": ",".join(map(str, lons)),
    }, timeout=60)
    r.raise_for_status()
    z = r.json()["elevation"]                      # metres
    dzdy = (z[1] - z[2]) / 200
    dzdx = (z[3] - z[4]) / 200
    slope = np.degrees(np.arctan(np.hypot(dzdx, dzdy)))
    terrain.append({"Sensor ID": s["Sensor ID"], "elevation_m": z[0], "slope_deg": round(slope, 2)})
terrain = pd.DataFrame(terrain)
terrain.to_csv(f"{raw_dir}/nyc_sensor_terrain.csv", index=False)
print(terrain.describe())
```

```python
import os, ast
import numpy as np
import pandas as pd

raw_dir = os.path.abspath("data/raw")

ev = pd.read_csv(f"{raw_dir}/floodnet_events.csv")
terr = pd.read_csv(f"{raw_dir}/nyc_sensor_terrain.csv")
ev = ev[ev["Sensor ID"].isin(terr["Sensor ID"])].copy()      # keep our 27 sensors

fmt = "%m/%d/%Y %I:%M:%S %p"
ev["start"] = pd.to_datetime(ev["Flood Start Datetime (GMT)"], format=fmt, utc=True)
ev["end"] = pd.to_datetime(ev["Flood End Datetime (GMT)"], format=fmt, utc=True)
ev["depths"] = ev["Time Series Depth Values (inches)"].apply(ast.literal_eval)
ev["secs"] = ev["Time Series Depth Timestamps (seconds)"].apply(ast.literal_eval)

print("Events:", len(ev), "| Sensors:", ev["Sensor ID"].nunique())
print("Duplicate events:", ev.duplicated(["Sensor ID", "start"]).sum())
print("End before start:", (ev["end"] <= ev["start"]).sum())
print("Depth/time lists different lengths:", (ev["depths"].str.len() != ev["secs"].str.len()).sum())

all_depths = np.concatenate(ev["depths"].values)
print("Negative depths:", (all_depths < 0).sum(), "| Max depth (in):", all_depths.max())

ev = ev.sort_values(["Sensor ID", "start"])
ev["prev_end"] = ev.groupby("Sensor ID")["end"].shift()
print("Overlapping events:", (ev["start"] < ev["prev_end"]).sum())

gaps = ev["secs"].apply(lambda s: np.diff(s).max() if len(s) > 1 else 0)
print("Biggest gap inside one event (min):", round(gaps.max() / 60, 1))
print("Events with a gap > 15 min:", (gaps > 900).sum())

rain = pd.read_csv(f"{raw_dir}/nyc_rain_hourly_openmeteo.csv", parse_dates=["time"])
print("\nRain negatives:", (rain["precipitation"] < 0).sum(), "| Max hourly rain (mm):", rain["precipitation"].max())
print("Rain hours per sensor:", rain.groupby("Sensor ID").size().unique())
```

```python
proc_dir = os.path.abspath("data/processed")
os.makedirs(proc_dir, exist_ok=True)

se = pd.read_csv(f"{raw_dir}/floodnet_sensors.csv").drop_duplicates("Sensor ID").set_index("Sensor ID")
hours = pd.date_range("2020-10-01 00:00", "2026-10-31 23:00", freq="h", tz="UTC")

out = []
for sid, g in ev.groupby("Sensor ID"):
    parts, filled = [], []
    for _, e in g.iterrows():
        secs, dep = np.array(e["secs"]), np.array(e["depths"])
        # minute-by-minute depth, straight-line between readings
        mins = pd.date_range(e["start"].ceil("min"), e["end"].floor("min"), freq="min")
        x = (mins - e["start"]).total_seconds()
        parts.append(pd.Series(np.interp(x, secs, dep), index=mins))
        parts.append(pd.Series(dep, index=(e["start"] + pd.to_timedelta(secs, unit="s")).floor("min")))
        # mark minutes that sit inside a gap of more than 15 minutes
        for a, b in zip(secs[:-1], secs[1:]):
            if b - a > 900:
                filled.append(pd.date_range((e["start"] + pd.Timedelta(seconds=a)).ceil("min"),
                                            (e["start"] + pd.Timedelta(seconds=b)).floor("min"), freq="min"))

    s = pd.concat(parts)
    s = s.groupby(s.index).max()                        # merges overlapping events
    depth = s.resample("h", label="right", closed="right").max().reindex(hours).fillna(0)

    flag = pd.Series(False, index=hours)
    if filled:
        fm = pd.Series(1, index=pd.DatetimeIndex(np.concatenate(filled)))
        fh = fm.resample("h", label="right", closed="right").max().reindex(hours).fillna(0)
        flag = fh > 0

    # blank out hours when the sensor wasn't installed
    installed = min(pd.Timestamp(se.loc[sid, "Date Installed"], tz="UTC"), g["start"].min().floor("D"))
    depth[hours < installed] = np.nan
    removed = se.loc[sid, "Date Removed"]
    if pd.notna(removed):
        depth[hours > pd.Timestamp(removed, tz="UTC") + pd.Timedelta(days=1)] = np.nan

    out.append(pd.DataFrame({"Sensor ID": sid, "time": hours, "depth_in": depth.values, "gap_filled": flag.values}))

depth_hourly = pd.concat(out, ignore_index=True)
depth_hourly.to_csv(f"{proc_dir}/nyc_depth_hourly.csv", index=False)

summary = depth_hourly.groupby("Sensor ID").agg(
    hours_observed=("depth_in", "count"),
    flooded_hours=("depth_in", lambda d: (d > 0).sum()),
    max_depth_in=("depth_in", "max"),
    gap_filled_hours=("gap_filled", "sum"),
).sort_values("flooded_hours", ascending=False)
print(summary.to_string())
print("\nTotal rows:", len(depth_hourly))
```

```python
import os
import requests
import pandas as pd

raw_dir = os.path.abspath("data/raw")
BASE = "https://data.cityofnewyork.us"

def save(url, name, params=None):
    try:
        with requests.get(url, params=params, stream=True, timeout=600) as r:
            r.raise_for_status()
            with open(f"{raw_dir}/{name}", "wb") as f:
                for chunk in r.iter_content(1 << 20):
                    f.write(chunk)
        print(f"OK   {name}: {os.path.getsize(f'{raw_dir}/{name}') / 1e6:.1f} MB")
        return True
    except Exception as e:
        print(f"FAIL {name}: {e}")
        return False

# 1. Flooding complaints since Oct 2020 (street flooding + clogged/flooding drains)
save(f"{BASE}/resource/erm2-nwe9.csv", "nyc_311_flooding.csv", params={
    "$select": "unique_key,created_date,closed_date,descriptor,incident_zip,"
               "incident_address,street_name,borough,latitude,longitude",
    "$where": "created_date >= '2020-10-01T00:00:00' AND descriptor in "
              "('Street Flooding (SJ)','Catch Basin Clogged/Flooding (Use Comments) (SC)')",
    "$limit": 500000,
})

# 2. Storm drains, 3. road segments, 4. stormwater flood maps
save(f"{BASE}/api/views/2w2g-fk3i/rows.csv?accessType=DOWNLOAD", "nyc_catch_basins.csv")
save(f"{BASE}/api/views/inkn-q76z/rows.csv?accessType=DOWNLOAD", "nyc_street_centerline.csv")
save(f"{BASE}/download/9i7c-xyvv/application/zip", "nyc_stormwater_flood_maps.zip")

# Quick look at what arrived
for name in ["nyc_311_flooding.csv", "nyc_catch_basins.csv", "nyc_street_centerline.csv"]:
    path = f"{raw_dir}/{name}"
    if os.path.exists(path):
        d = pd.read_csv(path, nrows=5)
        rows = sum(1 for _ in open(path, encoding="utf-8", errors="ignore")) - 1
        print(f"\n=== {name}: about {rows} rows ===")
        print(d.columns.tolist())
        print(d.head(2).to_string())

c = pd.read_csv(f"{raw_dir}/nyc_311_flooding.csv", usecols=["created_date", "descriptor"])
print("\nComplaints by type:\n", c["descriptor"].value_counts())
print("Dates:", c["created_date"].min(), "to", c["created_date"].max())
```

```python
import os, time
import pandas as pd
import requests

root = os.path.abspath("")
streets = pd.read_csv(f"{root}/data/processed/nyc_streets.csv")
out_path = f"{root}/data/raw/nyc_street_elevation.csv"

# resume if the cell was interrupted
done = pd.read_csv(out_path) if os.path.exists(out_path) else pd.DataFrame(columns=["street_id", "elevation_m"])
todo = streets[~streets["street_id"].isin(done["street_id"])]
print("Already done:", len(done), "| to download:", len(todo))

rows = []
for i in range(0, len(todo), 100):
    b = todo.iloc[i:i + 100]
    params = {"latitude": ",".join(b["mid_lat"].astype(str)),
              "longitude": ",".join(b["mid_lon"].astype(str))}
    for attempt in range(6):
        r = requests.get("https://api.open-meteo.com/v1/elevation", params=params, timeout=60)
        if r.status_code == 429:            # too many requests: wait, then retry
            print("Rate limit hit, waiting 60 s...")
            time.sleep(60)
            continue
        r.raise_for_status()
        break
    else:
        print("Stopped: still rate-limited. Run the cell again later; it resumes where it left off.")
        break
    rows.append(pd.DataFrame({"street_id": b["street_id"].values, "elevation_m": r.json()["elevation"]}))
    if len(rows) % 50 == 0:                 # save progress every 5,000 streets
        done = pd.concat([done] + rows, ignore_index=True); rows = []
        done.to_csv(out_path, index=False)
        print("Saved:", len(done))
    time.sleep(0.4)

done = pd.concat([done] + rows, ignore_index=True)
done.to_csv(out_path, index=False)
print("Finished:", len(done), "of", len(streets), "| missing:", done["elevation_m"].isna().sum())
print(done["elevation_m"].describe().round(1))
```

```python
import io, os, time, requests, pandas as pd

root = os.path.abspath("")
out_dir = f"{root}/data/raw/asos_1min"
os.makedirs(out_dir, exist_ok=True)
url = "https://mesonet.agron.iastate.edu/cgi-bin/request/asos1min.py"

months = pd.date_range("2020-10-01", "2026-11-01", freq="MS")
for a, b in zip(months[:-1], months[1:]):
    path = f"{out_dir}/{a:%Y-%m}.csv.gz"
    if os.path.exists(path):
        continue                                  # already downloaded: skip (lets you resume)
    for attempt in range(4):
        try:
            r = requests.get(url, timeout=600, params={
                "station": ["NYC", "JFK", "LGA"], "vars": "precip", "sample": "1min",
                "sts": f"{a:%Y-%m-%d}T00:00Z", "ets": f"{b:%Y-%m-%d}T00:00Z",
                "what": "download", "tz": "UTC"})
            r.raise_for_status()
            break
        except Exception as e:
            print(f"{a:%Y-%m} attempt {attempt + 1} failed: {e}")
            time.sleep(30)
    else:
        print("Stopped. Run the cell again later; it resumes where it left off.")
        break
    d = pd.read_csv(io.StringIO(r.text))
    d.to_csv(path, index=False, compression="gzip")
    print(f"{a:%Y-%m}: {len(d):,} rows")
    time.sleep(2)

# quick check of everything downloaded
files = sorted(os.listdir(out_dir))
d = pd.concat([pd.read_csv(f"{out_dir}/{f}") for f in files], ignore_index=True)
d["precip"] = pd.to_numeric(d["precip"], errors="coerce")       # 'M' = missing reading
d["valid(UTC)"] = pd.to_datetime(d["valid(UTC)"])
print("\nMonths:", len(files), "| rows:", f"{len(d):,}")
print("Missing readings by station:\n", d.groupby("station")["precip"].apply(lambda s: s.isna().mean()).round(3))
print("Total rain by station and year (inches):")
print(d.groupby(["station", d["valid(UTC)"].dt.year])["precip"].sum().unstack().round(1))
sep = d[d["valid(UTC)"].dt.date == pd.Timestamp("2023-09-29").date()]
print("\n29 Sep 2023 total (inches):\n", sep.groupby("station")["precip"].sum().round(2))
```

```python
import os, json, requests, pandas as pd

r = requests.get("https://archive-api.open-meteo.com/v1/archive", timeout=120, params={
    "latitude": 40.71, "longitude": -73.95, "start_date": "2026-01-01", "end_date": "2026-10-31",
    "hourly": "temperature_2m,relative_humidity_2m,wind_speed_10m,wind_gusts_10m,surface_pressure,cloud_cover",
    "timezone": "America/New_York"})
r.raise_for_status()
h = pd.DataFrame(r.json()["hourly"])
h["date"] = h["time"].str[:10]
names = {"temperature_2m": "temp", "relative_humidity_2m": "humidity", "wind_speed_10m": "wind",
         "wind_gusts_10m": "gust", "surface_pressure": "pressure", "cloud_cover": "cloud"}
out = {d: {v: g[k].round(1).tolist() for k, v in names.items()} for d, g in h.groupby("date")}
path = os.path.abspath("backend/data/weather.json")
json.dump(out, open(path, "w"))
print("Saved", len(out), "days to", path)
```
